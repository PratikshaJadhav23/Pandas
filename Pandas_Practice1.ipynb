```python
import pandas as pd

data = {
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Vikas",
             "Neha", "Arjun", "Pooja", "Karan", "Anjali"],
    "Department": ["IT", "HR", "IT", "Finance", "HR",
                   "Finance", "IT", "Marketing", "Marketing", "HR"],
    "Age": [25, 29, 32, 26, 35, 28, 24, 31, 27, 30],
    "Salary": [40000, 55000, 65000, 45000, 60000,
               50000, 35000, 70000, 48000, 52000],
    "Experience": [2, 5, 8, 3, 10, 4, 1, 7, 3, 6],
    "City": ["Mumbai", "Pune", "Mumbai", "Nashik", "Pune",
             "Mumbai", "Nashik", "Pune", "Mumbai", "Nashik"]
}

df = pd.DataFrame(data)

print(df)
```

Level 2: Filtering data (11–20)

```python
# 11. Find all employees whose salary is greater than 50,000
print(df[df["Salary"]> 50000]);
```

```python
# 12. Find employees whose age is less than 30.
print(df[df["Age"] < 30]);
```

```python
# 13. Find employees who work in the IT department.
print(df[df["Department"] == "IT"]);
```

```python
# 14. Find employees who live in Mumbai.
print(df[df["City"]=="Mumbai"]);
```

```python
# 15. Find employees whose salary is between 40,000 and 60,000, inclusive.
print(df[df["Salary"].between(40000,60000)])
```

```python
# 16. Find employees with more than 5 years of experience.
print(df[df["Experience"] > 5]);
```

```python
# 17. Find employees who work in HR and earn more than 50,000.
print(df[(df["Department"] == "HR") & (df["Salary"] > 50000)]);
```

```python
# 18. Find employees who live in Pune or Nashik.
print(df[df["City"].isin(["Pune","Nashik"])]);
```

```python
# 19. Find employees whose age is greater than 25 and salary is less than 60,000.
print(df[(df["Age"] > 25) & (df["Salary"] < 60000)]);
```

```python
# 20. Find employees who do not work in the IT department.
print(df[df["Department"] != "IT"]);
```
